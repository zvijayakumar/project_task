```python
import requests
from bs4 import BeautifulSoup
import json

# Base URL of the site
base_url = "https://subastas.boe.es"

# URL of the main page
main_url = "https://subastas.boe.es/index.php"

def scrape_auctions(url):
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, 'html.parser')

    auctions = []
    auction_list = soup.find('div', class_='listadoResult').find_all('li', class_='resultado-busqueda')

    for auction in auction_list:
        auction_details = {}
        auction_details['Title'] = auction.find('h3').text.strip()
        auction_details['Court'] = auction.find('h4').text.strip()
        auction_details['Expediente'] = auction.find('p').text.strip()
        auction_details['Estado'] = auction.find_all('p')[1].text.strip()
        if len(auction.find_all('p')) >= 3:
            auction_details['Detalles'] = auction.find_all('p')[2].text.strip()
        else:
            auction_details['Detalles'] = "NA"  # Or any default value
        auction_details['Link'] = "https://subastas.boe.es" + auction.find('a')['href'].strip()

        auctions.append(auction_details)

    return auctions

# Function to get area href links from the main page
def get_area_links(url):
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, 'html.parser')

    area_links = []
    map_element = soup.find('map', {'name': 'map'})
    
    if map_element is None:
        print("Map element with name 'map' not found!")
        return area_links

    map_areas = map_element.find_all('area')
    for area in map_areas:
        area_links.append(base_url + area['href'].strip())
    
    return area_links

# Function to get auction links from pagination
def get_pagination_links(soup):
    pagination = soup.find('div', class_='paginar2')
    if pagination:
        links = pagination.find_all('a', string=lambda text: text and 'siguiente' in text.lower())
        return [base_url + link['href'] for link in links if 'href' in link.attrs]
    return []

# Main script
all_auctions = []
area_links = get_area_links(main_url)

for link in area_links:
    print(f"Scraping URL: {link}")
    current_url = link
    while current_url:
        auctions = scrape_auctions(current_url)
        all_auctions.extend(auctions)

        # Get next page URL
        response = requests.get(current_url)
        response.raise_for_status()
        soup = BeautifulSoup(response.content, 'html.parser')
        pagination_links = get_pagination_links(soup)
        if pagination_links:
            next_page = pagination_links[-1]  # Get the last page link
            if next_page != current_url:
                current_url = next_page
            else:
                break
        else:
            break

# Save data to JSON file
with open('auctions.json', 'w', encoding='utf-8') as f:
    json.dump(all_auctions, f, ensure_ascii=False, indent=4)

print(f"Scraped {len(all_auctions)} auctions and saved to auctions.json")
```

```python
import requests
from bs4 import BeautifulSoup
import json
import time

# Base URL of the site
base_url = "https://subastas.boe.es/"

# URL of the page to scrape, with a limit of 500 results per page
url = ("https://subastas.boe.es/subastas_ava.php?campo%5B2%5D=SUBASTA.ESTADO.CODIGO&dato%5B2%5D=EJ"
       "&campo%5B3%5D=BIEN.TIPO&dato%5B3%5D=I&campo%5B8%5D=BIEN.COD_PROVINCIA&dato%5B8%5D=22"
       "&campo%5B17%5D=SUBASTA.FECHA_INICIO&dato%5B17%5D%5B0%5D=&dato%5B17%5D%5B1%5D=&page_hits=40"
       "&sort_field%5B0%5D=SUBASTA.FECHA_FIN&sort_order%5B0%5D=desc&accion=Buscar")
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.5",
    "Accept-Encoding": "gzip, deflate",
    "Referer": "https://www.google.com/"
}

# Function to scrape auction details from a given URL
def scrape_auctions(url):
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, 'html.parser')
    auctions = []
    auction_list = soup.find('div', class_='listadoResult').find_all('li', class_='resultado-busqueda')

    for auction in auction_list:
        auction_details = {}
        auction_details['Title'] = auction.find('h3').text.strip()
        auction_details['Court'] = auction.find('h4').text.strip()
        auction_details['Expediente'] = auction.find('p').text.strip()
        auction_details['Estado'] = auction.find_all('p')[1].text.strip()
        if len(auction.find_all('p')) >= 3:
            auction_details['Detalles'] = auction.find_all('p')[2].text.strip()
        else:
            auction_details['Detalles'] = "NA"  # Or any default value
        auction_details['Link'] = base_url + auction.find('a')['href'].strip()

        auctions.append(auction_details)

    return auctions

# Function to get auction links from pagination
def get_pagination_links(soup):
    pagination = soup.find('div', class_='paginar2')
    if pagination:
        links = pagination.find_all('a')
        return [base_url + link['href'] for link in links if 'href' in link.attrs]
    return []

# Function to scrape details from each auction's detail page
def scrape_auction_details(url):
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, 'html.parser')
    details = {}

    # Extract data from the 'Informacion General' section
    info_general = {}
    table_div = soup.find('div', id='idBloqueDatos1')
    if table_div:
        rows = table_div.find_all('tr')
        for row in rows:
            header = row.find('th').text.strip()
            value = row.find('td').text.strip()
            info_general[header] = value

    details['Informacion General'] = info_general

    return details

# Main script
all_auctions = []
current_url = url

while current_url:
    print(f"Scraping URL: {current_url}")
    auctions = scrape_auctions(current_url)
    for auction in auctions:
        print(f"Scraping details for auction: {auction['Title']}")
        try:
            auction_details = scrape_auction_details(auction['Link'])
            auction.update(auction_details)
        except Exception as e:
            print(f"Error scraping {auction['Link']}: {e}")
        time.sleep(0.5)  # To avoid overwhelming the server with requests

    all_auctions.extend(auctions)

    # Get next page URL
    response = requests.get(current_url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, 'html.parser')
    pagination_links = get_pagination_links(soup)
    if pagination_links:
        next_page = pagination_links[-1]  # Get the last page link
        if next_page != current_url:
            current_url = next_page
        else:
            break
    else:
        break

# Save data to JSON file
with open('auctions_extended.json', 'w', encoding='utf-8') as f:
    json.dump(all_auctions, f, ensure_ascii=False, indent=4)

print(f"Scraped {len(all_auctions)} auctions and saved to auctions_extended.json")
```
